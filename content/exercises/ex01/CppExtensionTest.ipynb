# Test C++ extension of NGSolve

Keep `mymodule.cpp` next to this notebook and start JupyterLab from that directory in the activated local environment. Run all cells: both assertions should pass and a success message should appear. Native C++ compilation is unavailable in JupyterLite.

Additional documentation: [C++ extension tutorial](https://docu.ngsolve.org/latest/i-tutorials/unit-9.1-C%2B%2BFE/CppExtension.html).

NGSolve compiles the C++ source and loads the library as a Python module:

```python
from ngsolve.fem import CompilePythonModule
from pathlib import Path

m = CompilePythonModule(Path('mymodule.cpp'), init_function_name='mymodule')
```

```python
from netgen.geom2d import unit_square
from ngsolve import Mesh, H1

assert m.answer() == 42
mesh = Mesh(unit_square.GenerateMesh(maxh=0.2))
fes = H1(mesh, order=2)
assert m.ndof(fes) == fes.ndof
print('C++ extension test passed.')
```

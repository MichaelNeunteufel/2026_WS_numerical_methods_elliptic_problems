# NGSolve and WebGUI check

Run the following cell. A triangular mesh of the unit square should appear below it.

```python
from netgen.occ import OCCGeometry, Rectangle
from ngsolve import Mesh
from ngsolve.webgui import Draw

square = Rectangle(1, 1).Face()
mesh = Mesh(OCCGeometry(square, dim=2).GenerateMesh(maxh=0.2))
Draw(mesh);
```
